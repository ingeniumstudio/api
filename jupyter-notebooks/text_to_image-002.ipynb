```python
# https://ipython.readthedocs.io/en/9.1.0/api/generated/IPython.display.html#IPython.display.Image
from IPython.display import Image
```

```python
def show_image(data: dict):
    image_bytes = text_to_image(**data)
    return Image(data=image_bytes)
```

```python
text = """\
text:
  paddded,

  ATTENTION: remember NerdFonts!\
"""
```

```python
image_data = {
    "text": text,
    "padding": 10,
    "foreground_color": "#660000",
    "background_color": "white",
    "font_size": 12,
}

show_image(image_data)
```

```python
image_data = {
    "text": text,
    "padding": 10,
    "foreground_color": "lightblue",
    "background_color": "black",
    "font_size": 12,
}

show_image(image_data)
```
